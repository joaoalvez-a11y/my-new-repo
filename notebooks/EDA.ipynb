```python
# importando Bibliotecas e o arquivo CSV
import pandas as pd
import plotly.express as px

car_data = pd.read_csv('../vehicles.csv')
```

```python
car_data.head()
```

```python
car_data.info()
```

```python
car_data.describe()
```

```python
# Verificando valores ausentes
car_data.isnull().sum()
```

```python
# Correção de valores ausentes na coluna 'is_4wd'
car_data['is_4wd'] = car_data['is_4wd'].fillna(0)
car_data['is_4wd'].value_counts()
```

```python
# Validando total da coluna 'is_4wd'
total_4wd = car_data['is_4wd'].count()
print(total_4wd)
```

```python
# Correção do tipo de dado da coluna 'date_posted'
car_data['date_posted'] = pd.to_datetime(car_data['date_posted'])
```

```python
# Validando tipo de dado da coluna 'date_posted'
car_data.info()
```

```python
# correção de valores ausentes na coluna 'model_year'
car_data['model_year'] = car_data['model_year'].fillna(car_data.groupby('model')['model_year'].transform('median'))
```

```python
# Validando o total da coluna 'model_year'
total_model_year = car_data['model_year'].count()
print(total_model_year)

car_data['model_year'].describe()
  
```

```python
# correção de valores ausentes na coluna 'cylinders'
car_data['cylinders'] = car_data['cylinders'].fillna(car_data.groupby('model')['cylinders'].transform('median'))
```

```python
# validando o total da coluna 'cylinders'
total_cylinders = car_data['cylinders'].count()
print(total_cylinders)

car_data['cylinders'].describe()
```

```python
# correção de valores ausentes na coluna 'paint_color'
car_data['paint_color'] = car_data['paint_color'].fillna('Unknown')
```

```python
# validando o total da coluna 'paint_color'
total_paint_color = car_data['paint_color'].count()
print(total_paint_color)

car_data['paint_color'].describe()
```

```python
# correção de valores ausentes na coluna 'odometer'
car_data['vehicle_age'] = 2019 - car_data['model_year'] + 1

car_data['odometer'] = car_data['odometer'].fillna(car_data.groupby('vehicle_age')['odometer'].transform('median'))
```

```python
# validando o total da coluna 'odometer'
total_odometer = car_data['odometer'].count()
print(total_odometer)

car_data['odometer'].describe()
```

```python
# Analise de valores outliers
car_data[['price', 'model_year', 'odometer', 'days_listed']].describe()
```

```python
# Analise de valores outliers quantis
desc = car_data['price'].describe()

limites = [desc['min'], desc['25%'], desc['50%'], desc['75%'], desc['max']]

faixas = pd.cut(car_data['price'], bins=limites, include_lowest=True)

print(faixas.value_counts().sort_index())
```
